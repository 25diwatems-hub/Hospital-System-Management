```python
import os
import getpass
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import gradio as gr

sns.set_theme(style="whitegrid")
DATA_FILE = "diabetes - diabetes.csv"

# --- Keep your exact class structure here ---
class HospitalSystem:
    def __init__(self, data_path=DATA_FILE):
        self.data_path = data_path
        if os.path.exists(self.data_path):
            self.df = pd.read_csv(self.data_path)
        else:
            self.df = pd.DataFrame()

    def view_data(self, num_rows):
        if self.df.empty:
            return "No data loaded or file missing."
        return self.df.head(int(num_rows))

    def get_summary(self):
        if self.df.empty:
            return "No data available."
        return self.df.describe().to_string()

    def generate_correlation_matrix(self):
        fig, ax = plt.subplots(figsize=(8, 6))
        if not self.df.empty:
            sns.heatmap(self.df.corr(), annot=True, cmap="Blues", ax=ax)
            ax.set_title("Feature Correlation Heatmap")
        else:
            plt.text(0.5, 0.5, "No Data File Found", ha='center')
        plt.tight_layout()
        return fig


# Instantiate your app object
app = HospitalSystem()

# --- Direct Gradio Interface Mapping ---
with gr.Blocks(title="Hospital System") as demo:
    gr.Markdown("# 🏥 Hospital System Interface")

    with gr.Tab("Dataset Viewer"):
        rows_slider = gr.Slider(minimum=1, maximum=50, value=5, step=1, label="Number of rows to view")
        view_btn = gr.Button("Display Data")
        data_table = gr.DataFrame(label="Dataset Head")

        view_btn.click(fn=app.view_data, inputs=rows_slider, outputs=data_table)

    with gr.Tab("Summary Statistics"):
        summary_btn = gr.Button("Get Data Summary")
        summary_output = gr.Textbox(label="Summary Statistics", lines=10)

        summary_btn.click(fn=app.get_summary, inputs=[], outputs=summary_output)

    with gr.Tab("Visualizations"):
        plot_btn = gr.Button("Show Correlation Matrix")
        plot_output = gr.Plot(label="Heatmap")

        plot_btn.click(fn=app.generate_correlation_matrix, inputs=[], outputs=plot_output)

if __name__ == "__main__":
    demo.launch(share=True)
```
